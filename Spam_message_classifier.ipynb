```python
import pandas as pd

data = {
    "message": [
        "Congratulations! You won a free prize. Click now!",
        "Win a free iPhone by clicking this link",
        "URGENT! You have won a cash reward",
        "Get free money now, claim your prize",
        "You have been selected for a special offer",
        "Claim your free gift today",
        "Free entry in our lucky draw",
        "Congratulations, you won a lottery",
        "Click here to win a free voucher",
        "You have won a free ticket",

        "Hey, are we meeting today?",
        "Please send me the assignment",
        "Can you call me when you are free?",
        "Your class starts at 10 AM tomorrow",
        "Don't forget to bring your notebook",
        "I will reach college in 15 minutes",
        "Can you share the project file?",
        "Let's meet after class",
        "Thank you for your help",
        "Are you coming to college today?"
    ],

    "label": [
        "Spam", "Spam", "Spam", "Spam", "Spam",
        "Spam", "Spam", "Spam", "Spam", "Spam",
        "Not Spam", "Not Spam", "Not Spam", "Not Spam", "Not Spam",
        "Not Spam", "Not Spam", "Not Spam", "Not Spam", "Not Spam"
    ]
}

df = pd.DataFrame(data)

print("Dataset created successfully!")
print()
print(df)
# Step 2: Text Preprocessing and Model Training

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report

# Convert text into numerical features
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(df["message"])

# Target labels
y = df["label"]

# Split dataset into training and testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

# Train the machine learning model
model = MultinomialNB()
model.fit(X_train, y_train)

# Make predictions
y_pred = model.predict(X_test)

# Evaluate model
accuracy = accuracy_score(y_test, y_pred)

print("Model Training Completed!")
print("Accuracy:", accuracy * 100, "%")
print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))
# Step 3: Spam Message Prediction

def predict_message(message):

    message_vector = vectorizer.transform([message])

    prediction = model.predict(message_vector)[0]

    if prediction == "Spam":
        print("🚨 Result: SPAM MESSAGE")
    else:
        print("✅ Result: NOT SPAM")

    print("Your Message:", message)


# Test messages
predict_message("Congratulations! You won a free prize")
predict_message("Hey, are we meeting today?")
predict_message("Claim your free cash reward now")
predict_message("Please send me the assignment")
```

```python
# STEP 4: Load Real SMS Spam Dataset

import pandas as pd
import zipfile
import io
import urllib.request

from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report

# Download official UCI dataset
url = "https://archive.ics.uci.edu/static/public/228/sms+spam+collection.zip"

response = urllib.request.urlopen(url)
zip_file = zipfile.ZipFile(io.BytesIO(response.read()))

# Read dataset
with zip_file.open("SMSSpamCollection") as file:
    df = pd.read_csv(
        file,
        sep="\t",
        header=None,
        names=["label", "message"]
    )

# Remove duplicate messages
df = df.drop_duplicates(subset="message").reset_index(drop=True)

# Convert labels
df["label"] = df["label"].map({
    "ham": "Not Spam",
    "spam": "Spam"
})

print("Dataset loaded successfully!")
print("Total messages:", len(df))

print("\nClass distribution:")
print(df["label"].value_counts())

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    df["message"],
    df["label"],
    test_size=0.20,
    random_state=42,
    stratify=df["label"]
)

# Create ML pipeline
model = Pipeline([
    ("tfidf", TfidfVectorizer(stop_words="english")),
    ("classifier", MultinomialNB())
])

# Train model
model.fit(X_train, y_train)

# Predictions
y_pred = model.predict(X_test)

# Evaluation
accuracy = accuracy_score(y_test, y_pred)

print("\nModel Training Completed!")
print("Accuracy:", round(accuracy * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```
